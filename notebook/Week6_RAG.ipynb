```python
from langchain_core.prompts import PromptTemplate

# rebuild chain with updated prompt
rag_chain = prompt_template | llm

def clean_summary(text: str) -> str:
    """
    Remove common Llama3 prefixes and suffixes that
    appear despite prompt instructions.
    """
    # remove common prefixes
    prefixes = [
        "Here is the market intelligence briefing:",
        "Here is the briefing:",
        "MARKET INTELLIGENCE BRIEFING:",
        "BRIEFING:",
        "Market Intelligence Briefing:",
    ]
    for prefix in prefixes:
        if text.strip().startswith(prefix):
            text = text.strip()[len(prefix):].strip()

    # remove markdown bold markers
    text = text.replace("**Market Sentiment and Trend:**", "")
    text = text.replace("**Significant Event/Ticker:**", "")
    text = text.replace("**Most Significant Specific Event:**", "")
    text = text.strip()

    return text

print("RAG chain rebuilt with cleaner prompt ")
print("clean_summary() function defined ")
```

```python
import time

def generate_market_summary(query: str, top_k: int = 5) -> dict:
    """
    Complete RAG pipeline:
    1. Search Qdrant for top_k similar headlines
    2. Build context from retrieved headlines
    3. Generate 2-sentence summary using Llama3

    Args:
        query:  user query string
        top_k:  number of headlines to retrieve as context

    Returns:
        dictionary with summary, sources and timing
    """
    if not query or not query.strip():
        raise ValueError("Query cannot be empty")

    start_time = time.time()

    # step 1: retrieve similar headlines from Qdrant
    print(f"Step 1: Retrieving top {top_k} similar headlines...")
    search_results = qdrant_search(
        client          = qdrant_client,
        query           = query,
        top_k           = top_k,
        min_similarity  = 0.3,
        collection_name = "financial_headlines"
    )

    if not search_results:
        return {
            "query":   query,
            "summary": "No relevant headlines found for this query.",
            "sources": [],
            "processing_time_ms": round(
                (time.time() - start_time) * 1000, 2
            )
        }

    retrieval_time = (time.time() - start_time) * 1000
    print(f"  Retrieved {len(search_results)} headlines in "
          f"{retrieval_time:.1f}ms")

    # step 2: build context string
    print("Step 2: Building context...")
    context_lines = []
    for i, result in enumerate(search_results):
        context_lines.append(
            f"{i+1}. [{result['sentiment']}] "
            f"{result['headline']} "
            f"(similarity: {result['similarity']:.2f})"
        )
    context = "\n".join(context_lines)

    print(f"  Context built from {len(search_results)} headlines")

    # step 3: generate summary using Llama3
    print("Step 3: Generating summary with Llama3...")
    llm_start = time.time()

    summary = rag_chain.invoke({
        "context":       context,
        "num_headlines": len(search_results)
    })

    summary = clean_summary(summary)

    llm_time = (time.time() - llm_start) * 1000
    print(f"  Summary generated in {llm_time:.1f}ms")

    total_time = (time.time() - start_time) * 1000

    return {
        "query":   query,
        "summary": summary.strip(),
        "sources": [
            {
                "rank":       r["rank"],
                "similarity": r["similarity"],
                "headline":   r["headline"],
                "sentiment":  r["sentiment"],
                "ticker":     r["ticker"]
            }
            for r in search_results
        ],
        "retrieval_time_ms": round(retrieval_time, 2),
        "llm_time_ms":       round(llm_time, 2),
        "processing_time_ms": round(total_time, 2)
    }

print("generate_market_summary() defined successfully")
```

```python
def generate_sentiment_summary(query: str,
                                sentiment: str,
                                top_k: int = 5) -> dict:
    """
    RAG with sentiment filter —
    generates summary from only Bearish or only Bullish headlines.
    """
    search_results = qdrant_search(
        client           = qdrant_client,
        query            = query,
        top_k            = top_k,
        min_similarity   = 0.3,
        sentiment_filter = sentiment,
        collection_name  = "financial_headlines"
    )

    if not search_results:
        return {
            "query":    query,
            "sentiment_filter": sentiment,
            "summary":  f"No {sentiment} headlines found for this query.",
            "sources":  []
        }

    context = "\n".join([
        f"{i+1}. [{r['sentiment']}] {r['headline']}"
        for i, r in enumerate(search_results)
    ])

    summary = rag_chain.invoke({
        "context":       context,
        "num_headlines": len(search_results)
    })

    return {
        "query":            query,
        "sentiment_filter": sentiment,
        "summary":          summary.strip(),
        "sources":          search_results
    }

# test bearish summary
print("=== Bearish Market Summary ===\n")
result = generate_sentiment_summary(
    "oil prices economy global markets",
    sentiment="Bearish",
    top_k=5
)
print(f"Filter: Bearish only")
print(f"Sources: {len(result['sources'])} headlines")
print(f"\nBEARISH MARKET BRIEFING:")
print(result["summary"])

print("\n=== Bullish Market Summary ===\n")
result = generate_sentiment_summary(
    "technology stocks earnings growth",
    sentiment="Bullish",
    top_k=5
)
print(f"Filter: Bullish only")
print(f"Sources: {len(result['sources'])} headlines")
print(f"\nBULLISH MARKET BRIEFING:")
print(result["summary"])
```
